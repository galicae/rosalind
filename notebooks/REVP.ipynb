```python
from rutil import reverse_complement
from rio import read_fasta
```

```python
def is_rev_palindrome(dna):
    return dna == reverse_complement(dna)
```

```python
assert is_rev_palindrome("GCATGC")
```

```python
def find_reverse_palindromes(dna, palindrome_min=4, palindrome_max=12):
    for l in range(palindrome_min, palindrome_max+1, 2):
        for i in range(len(dna) - l + 1):
            if is_rev_palindrome(dna[i:i+l]):
                print(i+1, l)
```

```python
dna = "TCAATGCATGCGGGTCTATATGCAT"
find_reverse_palindromes(dna)
```
